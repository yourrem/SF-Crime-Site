# Step 2 — Explore Incident Reports Dataset
Load raw API data into pandas and inspect columns, types, and nulls.

```python
import sys
sys.path.append('..')

import pandas as pd
from etl.extract import fetch_incidents
```

```python
raw = fetch_incidents(days=7)
df = pd.DataFrame(raw)
print(f'Shape: {df.shape}')
df.head()
```

```python
# Column names and dtypes
df.dtypes
```

```python
# Null counts per column
df.isnull().sum().sort_values(ascending=False)
```

```python
# Quick summary stats
df.describe(include='all')
```

```python
# Unique values in key categorical columns
for col in ['incident_category', 'incident_subcategory', 'resolution', 'police_district']:
    if col in df.columns:
        print(f'\n--- {col} ---')
        print(df[col].value_counts())
```
